```python
from pyspark.sql import functions as F

# Products - product_id ist unique
df_products = spark.read.table("lh_bronze_olist.dbo.olist_products_dataset")

df_products_clean = (
    df_products
    .dropDuplicates(["product_id"])
)

df_products_clean.write.format("delta").mode("overwrite") \
    .option("overwriteSchema", "true").saveAsTable("lh_silver_olist.dbo.products")

print(f"products: {df_products_clean.count()} records")
```

```python
# Category Translation - expect no dupplication, but as a precaution
df_translation = spark.read.table("lh_bronze_olist.dbo.product_category_name_translation")

df_translation_clean = (
    df_translation
    .dropDuplicates(["product_category_name"])
)

df_translation_clean.write.format("delta").mode("overwrite") \
    .option("overwriteSchema", "true").saveAsTable("lh_silver_olist.dbo.product_category_translation")

print(f"category_translation: {df_translation_clean.count()} records")
```
