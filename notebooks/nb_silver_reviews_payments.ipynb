```python
from pyspark.sql import functions as F

# Order Items - composite key!
df_items = spark.read.table("lh_bronze_olist.dbo.olist_order_items_dataset")

df_items_clean = (
    df_items
    .withColumn("shipping_limit_date", F.to_timestamp("shipping_limit_date"))
    .dropDuplicates(["order_id", "order_item_id"])
)

df_items_clean.write.format("delta").mode("overwrite") \
    .option("overwriteSchema", "true").saveAsTable("lh_silver_olist.dbo.order_items")

print(f"order_items: {df_items_clean.count()} records")
```

```python
# Payments - composite key!
df_payments = spark.read.table("lh_bronze_olist.dbo.olist_order_payments_dataset")

df_payments_clean = (
    df_payments
    .dropDuplicates(["order_id", "payment_sequential"])
)

df_payments_clean.write.format("delta").mode("overwrite") \
    .option("overwriteSchema", "true").saveAsTable("lh_silver_olist.dbo.payments")

print(f"payments: {df_payments_clean.count()} records")
```

```python
# Reviews - composite key!
df_reviews = spark.read.table("lh_bronze_olist.dbo.olist_order_reviews_dataset")

df_reviews_clean = (
    df_reviews
    .withColumn("review_creation_date", F.to_timestamp("review_creation_date"))
    .withColumn("review_answer_timestamp", F.to_timestamp("review_answer_timestamp"))
    .dropDuplicates(["review_id", "order_id"])
)

df_reviews_clean.write.format("delta").mode("overwrite") \
    .option("overwriteSchema", "true").saveAsTable("lh_silver_olist.dbo.reviews")

print(f"reviews: {df_reviews_clean.count()} records")
```
